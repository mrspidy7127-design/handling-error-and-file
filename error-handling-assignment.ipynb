### Q5
`numbers.txt` file ko read karein aur tamam numbers ka sum (total) calculate karke print karein.

```python
# Q5 solution code here
tot_sum = 0
with open("number.txt","r") as f:
    for s in f:
        clean_number = int(s.strip())

        tot_sum += clean_number

print(f"The sum of all number is: {tot_sum}")
```

### Q17
Ek `try-except` block likhein jo Multiple Exceptions catch kar sake (`ZeroDivisionError` aur `ValueError`).

```python
# Q17 solution code here
try:
    num = int(input("Enter a number"))
    result = 10/num
    print("Ans:",result)
except ZeroDivisionError:
    print("You cannot divide a number by zero")
except ValueError:
    print("Error: Invalid Input! Meherbani karke sirf number type karein")
```

### Q18
`else` block ka istemal `try-except` ke sath karein jo sirf tab chale jab koi error na aaye.

```python
# Q18 solution code here

try:
    num = int(input("Enter a number: "))
    result = 10 / num

except ValueError:
   
    print("Error: Invalid Input! Meherbani karke sirf ek number type karein.")

except ZeroDivisionError:
    
    print("Error: Aap kisi number ko 0 se divide nahi kar sakte!")

else:
    
    print("Mubarak ho! Code bina kisi error ke chal gaya.")
    print(f"Aapka final jawab hai: {result}")
```

### Q19 
`raise` keyword ka use karke age validation ka function banayein. Agar age 18 se kam ho toh custom `ValueError("Age must be 18 or older")` raise karein.

```python
# Q19 solution code here
def validate_age(age):
    if age < 18:
         raise ValueError("Age must be 18 or older")
    else:
        print("Mubarak ho! Aap eligible hain.")
try:
    user_age = int(input("Apni age enter karein: "))
    validate_age(user_age)

except ValueError as e:
    print(f"Validation Error: {e}")
```

### Q20 
File handling aur Exception handling ko combine karein: Ek file read karein, agar file na mile toh error catch karein, aur `finally` block mein file closure handle karein.

```python
# Q20 solution code here
f = None
try:
    f = open("Final.txt","r")
    print(f.read())
except FileNotFoundError:
     print("Error: File nahi mili! file ka naam check karein.")

finally:
    if f is not None:
        f.close()
        print("Finally Block: File ko safe tareeqe se close kar diya hai.")
    else:
        print("Finally Block: File open hi nahi hui thi, isliye close karne ki zaroorat nahi pari.")    
```
